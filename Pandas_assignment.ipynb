```python
# 14. Handling Missing Values
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

print(df)
print(df.isnull())
print(df.isnull().sum())
df = df.dropna()
print(df)

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)
df["Marks"] = df["Marks"].fillna(0)

print(df)

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, None, 19],
    "Marks": [80, 90, None]
}

df = pd.DataFrame(data)
print(df.isnull().sum())
```

```python
# 15. Handling Duplicate Data
import pandas as pd

data = {
    "Name": ["Ayesha", "Bilal", "Hamza", "Ayesha", "Sana"],
    "Marks": [85, 72, 90, 85, 78]
}

df = pd.DataFrame(data)
print(df.duplicated())
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df)
```

```python
# 16. Sorting Data
import pandas as pd

data = {
    "Name": ["Ayesha", "Hamza", "Sana", "Bilal", "Zain"],
    "Marks": [82, 65, 91, 74, 88]
}

df = pd.DataFrame(data)
print(df.sort_values("Marks"))
print(df.sort_values("Marks", ascending=False))
```

```python
# 17. Aggregation and GroupBy
import pandas as pd

data = {
    "Name": ["Ayesha", "Bilal", "Hamza", "Sana", "Zain"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Marks": [85, 72, 90, 68, 78]
}

df = pd.DataFrame(data)

marks = df["Marks"]

print(marks.sum())
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.count())
result = df.groupby("Department")["Marks"].mean()
print(result)
result = df.groupby("Department")["Marks"].sum()
print(result)
result = df.groupby("Department")["Marks"].count()
print(result)
```

```python
# 18. Combining DataFrames
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Hina", "Usman", "Maham"],
    "Marks": [78, 85, 92]
})

df2 = pd.DataFrame({
    "Name": ["Daniyal", "Iqra", "Fahad"],
    "Marks": [69, 88, 76]
})

final_df = pd.concat([df1, df2], ignore_index=True)

print(final_df)
```

```python
students = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Hina", "Usman", "Maham", "Daniyal"]
})

results = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Grade": ["B", "A", "A+", "C"]
})

final_df = pd.merge(students, results, on="ID")

print(final_df)
```
